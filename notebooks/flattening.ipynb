```python
df['lat'] = df['coordinates'].lat
```

```python
df['lat'] = df['coordinates'].apply(lambda d: d['lat'])
df['long'] = df['coordinates'].apply(lambda d: d['lng'])

```

```python
df.drop(columns='coordinates', inplace=True)
```
