# Extract the Data


```python
import requests
import pandas as pd
```

Here we first extract the first 100 countries then save it into json

```python
response = requests.get(
  'https://api.restcountries.com/countries/v5?limit=100',
  headers={'Authorization': 'Bearer rc_live_62ad5d16ca21438690937b3dacece8fb'}
).json()
```

```python
response['data']['meta']
```

From `response['data']['meta']` we can conclude that we have a total of 254 countries in the API.

```python
df1 = pd.DataFrame(response['data']['objects'])
```

Now we repeat for the next 100 countries.

```python
response = requests.get(
  'https://api.restcountries.com/countries/v5?limit=100&offset=100',
  headers={'Authorization': 'Bearer rc_live_62ad5d16ca21438690937b3dacece8fb'}
).json()

df2 = pd.DataFrame(response['data']['objects'])
```

Now we need the last 54 countries.

```python
response = requests.get(
  'https://api.restcountries.com/countries/v5?limit=100&offset=200',
  headers={'Authorization': 'Bearer rc_live_62ad5d16ca21438690937b3dacece8fb'}
).json()

df3 = pd.DataFrame(response['data']['objects'])
```

Now we concat all 3 dataframes and save it into a single json file.

```python
pd.concat([df1, df2, df3]).to_json('../data/raw/countries.json', orient='records')
```
